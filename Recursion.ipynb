```python
#Binary Search without recurion
A=[1,3,5,6,10,30,31,40,50]
def binary_search(A,x):
    up=len(A)-1
    down=0
    while(down<=up):
        mid = (up+down)//2
        if(x==A[mid]):
            return mid
        elif(x>=A[mid]):
            down=mid+1
        else:
            up=mid-1
    return False
print(binary_search(A,50))
    
```

```python
# Recursive Binary Search
A=[1,3,5,6,10,30,31,40,50]
def binary_search_recursive(A,x,down,up):
    if(down>up):
        return False
    mid =(up+down)//2
    if(A[mid]==x):
        return mid
    elif(A[mid]<x):
        return binary_search_recursive(A,x,mid+1,up)
    else:
        return binary_search_recursive(A,x,down,mid-1)
print(binary_search_recursive(A,50,0,8))
```

```python
#Merge Sort is of order of nlogn
A=[7,5,1, 6, 9, 10,4,2,30,1]

def merge_sort(A):
    if(len(A)<2):
        return A
    mid=len(A)//2
    Left = A[:mid]
    Right = A[mid:]
    Left = merge_sort(Left)
    Right = merge_sort(Right)
    A=[]
    i=0
    j=0
    while i<len(Left) and j<len(Right):
        if(Left[i]<=Right[j]):
            A.append(Left[i])
            i+=1
        elif(Left[i]>Right[j]):
            A.append(Right[j])
            j+=1
    A+=Left[i:]
    A+=Right[j:]
    return A
print(merge_sort(A))     
```

Master Theorem
</font>
<p></p>

If we have a recurrence relation of the following form:

$$T(n)=aT\left(\frac{n}{b}\right)+f(n)$$

then four possible cases may occur:

<ul>

<li>
There exists a number $\epsilon>0$ such that

$$f(n)\in O\left(n^{\log_b a-\epsilon}\right).$$

This means that $f(n)$ is polynomially smaller than $n^{\log_b a}$.
In this case, we have:

$$T(n)\in \Theta\left(n^{\log_b a}\right).$$
</li>

<li>
There exists a number $\epsilon>0$ such that

$$f(n)\in \Omega\left(n^{\log_b a+\epsilon}\right).$$

This means that $f(n)$ is polynomially larger than $n^{\log_b a}$.
In this case, we have:

$$T(n)\in \Theta(f(n)).$$
</li>

<li>
We have

$$n^{\log_b a}\in \Theta(f(n)).$$

This means that $f(n)$ and $n^{\log_b a}$ have the same asymptotic order.
In this case, we have:

$$T(n)\in \Theta(f(n)\lg n)
=
\Theta\left(n^{\log_b a}\lg n\right).$$
</li>

<li>
None of the above three cases holds. In this case, the Master Theorem cannot be used, and we must use other methods to find the complexity of the recurrence relation.
</li>

</ul>

## Mathematical Analysis: Binary Counter

A binary counter sometimes flips many bits in one increment. For example, `0111 → 1000` flips 4 bits, so the worst-case cost of one increment is `O(k)` for a `k`-bit counter.

But amortized analysis looks at many increments together. Over `n` increments, the last bit flips every time, the next bit flips every 2 times, the next every 4 times, and so on.

So the total number of bit flips is:

`n + n/2 + n/4 + n/8 + ... ≤ 2n`

That means `n` increments cost at most `O(n)` total, so the average cost per increment is:

`O(n) / n = O(1)`

**Takeaway:** One binary-counter increment can be expensive, but expensive increments are rare, so incrementing a binary counter takes **O(1) amortized time**.

## Coin Analysis: Binary Counter

Use the **coin method** by charging each increment **2 coins**.

When we flip the last bit from `0 → 1`, we pay **1 coin** for that flip and store **1 coin** on that bit.

Later, when that bit flips from `1 → 0`, we use the stored coin to pay for it.

So every `1 → 0` flip is already prepaid, and every increment only needs to pay directly for the new `0 → 1` flip.

Therefore, charging **2 coins per increment** pays for all bit flips.

So the amortized cost per increment is:

`O(1)`

**Takeaway:** Store a coin on each bit when it becomes `1`; use that coin when it turns back to `0`.

## Potential Method: Binary Counter

Use potential:

`Φ = number of 1 bits in the counter`

Think of each `1` as stored energy that can pay for flipping back to `0`.

When we increment, suppose there are `t` trailing `1`s.

Example:

`0111 -> 1000`

Here, `t = 3`.

The increment does:

- flip `t` ones to zero
- flip one zero to one

Actual cost:

`t + 1`

Potential change:

- losing `t` ones decreases potential by `t`
- gaining one new `1` increases potential by `1`

So:

`ΔΦ = 1 - t`

Amortized cost:

`actual cost + ΔΦ`

`= (t + 1) + (1 - t)`

`= 2`

**Takeaway:** Even if one increment flips many bits, the potential stored in the old `1`s pays for it. So increment is `O(1)` amortized.

## Coin Analysis: Dynamic Array Stack

For each `push`, we charge **3 coins**.

- **1 coin** pays for inserting the new element.
- **2 coins** are saved on the new element for future copying.

When the array becomes full, we double its size and copy all old elements.

Each old element already has saved coins on it, so those saved coins pay for moving it into the new array.

So even though one resize can cost `O(n)`, the cost was prepaid by earlier pushes.

Therefore, `n` pushes cost at most `3n` coins total.

Average cost per push:

`3n / n = O(1)`

**Takeaway:** Charge 3 coins per push: 1 pays for the push now, and 2 are saved to pay for future resizing. So `push` is **O(1) amortized**.
