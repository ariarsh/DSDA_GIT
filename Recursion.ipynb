# Recursion

```python
#Binary Search without recurion
A=[1,3,5,6,10,30,31,40,50]
def binary_search(A,x):
    up=len(A)-1
    down=0
    while(down<=up):
        mid = (up+down)//2
        if(x==A[mid]):
            return mid
        elif(x>=A[mid]):
            down=mid+1
        else:
            up=mid-1
    return False
print(binary_search(A,50))
    
```

```python
# Recursive Binary Search
A=[1,3,5,6,10,30,31,40,50]
def binary_search_recursive(A,x,down,up):
    if(down>up):
        return False
    mid =(up+down)//2
    if(A[mid]==x):
        return mid
    elif(A[mid]<x):
        return binary_search_recursive(A,x,mid+1,up)
    else:
        return binary_search_recursive(A,x,down,mid-1)
print(binary_search_recursive(A,50,0,8))
```

```python
#Merge Sort
A=[7,5,1, 6, 9, 10,4,2,30,1]

def merge_sort(A):
    if(len(A)<2):
        return A
    mid=len(A)//2
    Left = A[:mid]
    Right = A[mid:]
    Left = merge_sort(Left)
    Right = merge_sort(Right)
    A=[]
    i=0
    j=0
    while i<len(Left) and j<len(Right):
        if(Left[i]<=Right[j]):
            A.append(Left[i])
            i+=1
        elif(Left[i]>Right[j]):
            A.append(Right[j])
            j+=1
    A+=Left[i:]
    A+=Right[j:]
    return A
print(merge_sort(A))     
```
