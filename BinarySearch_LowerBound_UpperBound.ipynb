# Binsary Search Lower Bound Upper Bound

```python
A = [1,2,3,4,5,6,7,8,20,40,50,60,100,102,104]

def binary_search(A,x,l=0,r=len(A)):
    if(x<A[l] or A[r-1]<x):
        return False
    elif(l+1==r):
        if(A[l]==x):
            return True
        else:
            return False
    mid = (l+r)//2
    if(x<A[mid]):
        return binary_search(A,x,l,mid)
    else:
        return binary_search(A,x,mid,r)

binary_search(A,40)
```

```python
# Best Version
def binary_search(A,x):
    l,r = 0, len(A)
    
    while(r-l>1):
        mid = (l+r)//2
        if(x<A[mid]):
            r = mid
        else:
            l = mid

    return A[l]==x

binary_search(A,104)
```

# Lower bound: largest i in which A[i]<x
### binary search returns the first founded element, so in case of repeated elements, we cannot return l-1 for lower bound!

```python
def lower_bound(A,x):
    l,r =0,len(A)
    if(x<=A[0]):
        return -1

    while (r-l>1):
        mid = (l+r)//2
        if(x<=A[mid]):
            r=mid
        else:
            l=mid
    return (l,A[l])  

lower_bound(A,102)
```
