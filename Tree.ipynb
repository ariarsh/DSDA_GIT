# Removing an Item from a Binary Search Tree

## 1. Leaf Node

If the node has **no children**, simply remove it.

**Example:** Delete `5`.

```text
Before:     10
           /
          5

After:      10
```

---

## 2. Node with One Child

If the node has **one child**, replace the node with its child.

**Example:** Delete `5`, so `7` takes the place of `5`.

```text
Before:     10
           /
          5
           \
            7

After:      10
           /
          7
```

---

## 3. Node with Two Children

If the node has **two children**, replace it with its **inorder successor**, which is the smallest value in the right subtree. Then delete the successor node.
We can also bring biggest in the left subtree!

**Example:** Delete `10`; successor is `12`.

```text
Before:

              10
             /  \
            5    20
                /
               12
                 \
                  15
                    \
                     17
```

```text
After:

              12
             /  \
            5    20
                /
               15
                 \
                  17
```

````python
#make post oredr from in_order and pre_order
in_order = [4 ,2 ,5 ,1 ,3 ,6]
pre_order = [1 ,2 ,4 ,5 ,3 ,6]

def get_post_order(preorder, inorder):
    if len(preorder) == 0:
        return []

    root = preorder[0]

    root_index = inorder.index(root)

    left_inorder = inorder[:root_index]
    right_inorder = inorder[root_index + 1:]

    left_size = len(left_inorder)

    left_preorder = preorder[1:1 + left_size]
    right_preorder = preorder[1 + left_size:]

    left_postorder = get_post_order(left_preorder, left_inorder)
    right_postorder = get_post_order(right_preorder, right_inorder)

    return left_postorder + right_postorder + [root]


post_order = get_post_order(pre_order, in_order)
for i in range(len(post_order)):
    print (post_order[i],end=" ")


````

# Catalan Formula for Number of Binary Search Trees

The number of different binary search trees that can be made from `n` distinct sorted numbers is the **nth Catalan number**, denoted by:

$$
C_n
$$

## Recursive Formula

The base case is:

$$
C_0 = 1
$$

For $n \ge 1$:

$$
C_n = \sum_{i=0}^{n-1} C_i C_{n-1-i}
$$

This recurrence comes from choosing each number as the root.

If the root has:

- $i$ numbers on the left
- $n - 1 - i$ numbers on the right

then the number of trees for that root choice is:

$$
C_i C_{n-1-i}
$$

Summing over all possible root positions gives:

$$
C_n = \sum_{i=0}^{n-1} C_i C_{n-1-i}
$$

## Exact Formula

The closed-form formula for the Catalan number is:

$$
C_n = \frac{1}{n+1}\binom{2n}{n}
$$

Equivalently:

$$
C_n = \frac{(2n)!}{(n+1)!n!}
$$

## Deriving the Exact Formula from the Recurrence

Start with the recurrence:

$$
C_n = \sum_{i=0}^{n-1} C_i C_{n-1-i}
$$

Define the generating function:

$$
C(x) = \sum_{n=0}^{\infty} C_n x^n
$$

Using the recurrence, the generating function satisfies:

$$
C(x) = 1 + xC(x)^2
$$

Rearrange this into a quadratic equation:

$$
xC(x)^2 - C(x) + 1 = 0
$$

Using the quadratic formula:

$$
C(x) = \frac{1 \pm \sqrt{1 - 4x}}{2x}
$$

We choose the minus sign because we need $C(0) = 1$:

$$
C(x) = \frac{1 - \sqrt{1 - 4x}}{2x}
$$

Expanding this generating function gives:

$$
C(x) = \sum_{n=0}^{\infty} \frac{1}{n+1}\binom{2n}{n}x^n
$$

Therefore:

$$
C_n = \frac{1}{n+1}\binom{2n}{n}
$$
