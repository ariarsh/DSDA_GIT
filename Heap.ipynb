# Heap

# What data structure to save a priority list of patients?


| Data Structure | Insert | Find Max | Delete Max | Notes |
|---|---:|---:|---:|---|
| Unsorted Linked List | `O(1)` | `O(n)` | `O(n)` | Insert at head is constant time, but finding the max requires scanning the list. Once the max node is found, deleting it is `O(1)`. |
| Sorted Linked List — Ascending | `O(n)` | `O(1)` | `O(n)` for singly linked list, `O(1)` for doubly linked list with tail pointer | Max is at the end. In a singly linked list, deleting the last node requires finding the previous node. |
| Balanced BST | `O(log n)` | `O(log n)` | `O(log n)` | Tree height is `O(log n)`, so insert, find max, and delete are logarithmic. |
| Max Heap | `O(log n)` | `O(1)` | `O(log n)` | Max is always at the root. Insert and delete require heapifying. |

# Heap and Array Representation

A **binary heap** is a complete binary tree that is commonly stored as an array.

The array is formed by reading the heap **level by level**, from **left to right**.


### Example: Max Heap

```text
            90
          /    \
        75      60
       /  \    /  \
     40   50  30   20
    /
  10
```

### Corresponding Array

```text
Index:   0   1   2   3   4   5   6   7
Array: [90, 75, 60, 40, 50, 30, 20, 10]
```
