# Heap and quick_sort

# What data structure to save a priority list of patients?


| Data Structure | Insert | Find Max | Delete Max | Notes |
|---|---:|---:|---:|---|
| Unsorted Linked List | `O(1)` | `O(n)` | `O(n)` | Insert at head is constant time, but finding the max requires scanning the list. Once the max node is found, deleting it is `O(1)`. |
| Sorted Linked List — Ascending | `O(n)` | `O(1)` | `O(n)` for singly linked list, `O(1)` for doubly linked list with tail pointer | Max is at the end. In a singly linked list, deleting the last node requires finding the previous node. |
| Balanced BST | `O(log n)` | `O(log n)` | `O(log n)` | Tree height is `O(log n)`, so insert, find max, and delete are logarithmic. |
| Max Heap | `O(log n)` | `O(1)` | `O(log n)` | Max is always at the root. Insert and delete require heapifying. |

# Heap and Array Representation

A **binary heap** is a complete binary tree that is commonly stored as an array.

The array is formed by reading the heap **level by level**, from **left to right**.


### Example: Max Heap

```text
            90
          /    \
        75      60
       /  \    /  \
     40   50  30   20
    /
  10
```

### Corresponding Array

```text
Index:   0   1   2   3   4   5   6   7
Array: [90, 75, 60, 40, 50, 30, 20, 10]
```


### Index Rules

For a node stored at index `i`:

| Relationship | Formula |
|---|---|
| Left child | `2i + 1` |
| Right child | `2i + 2` |
| Parent | `(i - 1) // 2` |




````python
class MaxHeap:
    def __init__(self, lis=None):
        if lis is None:
            self.arr = []
        else:
            self.arr = lis.copy()
            self.build_heap()

    def parent_index(self, i):
        return (i - 1) // 2

    def left_child_index(self, i):
        return 2 * i + 1

    def right_child_index(self, i):
        return 2 * i + 2

    def insert(self, member):
        self.arr.append(member)
        self.bubble_up(len(self.arr)-1)
        
        
    def peek_max(self):
        if(len(self.arr)==0):
            return None
        else:
            return self.arr[0]

    def delete_max(self):
        if len(self.arr) == 0:
            return None   # or raise an error
    
        max_value = self.arr[0]
    
        if len(self.arr) == 1:
            self.arr.pop()
            return max_value
    
        self.arr[0] = self.arr[-1]
        self.arr.pop()
        self.bubble_down(0)
    
        return max_value

    def bubble_up(self, index):
        while(index>0):
            parent = self.parent_index(index)
            if(self.arr[parent]<self.arr[index]):
                self.arr[parent],self.arr[index] = self.arr[index], self.arr[parent]
                index = parent
            else:
                break
            

    def bubble_down(self, index):
        while self.left_child_index(index) < len(self.arr):
            left = self.left_child_index(index)
            right = self.right_child_index(index)
    
            larger_child = left
    
            if right < len(self.arr) and self.arr[right] > self.arr[left]:
                larger_child = right
    
            if self.arr[larger_child] > self.arr[index]:
                self.arr[larger_child], self.arr[index] = self.arr[index], self.arr[larger_child]
                index = larger_child
            else:
                break
            

    def build_heap(self):  # this is O(nlogn)
        for i in range(len(self.arr)):
            self.bubble_up(i)

    def build_heap_bubbledown(self): # this is O(n)
        last_non_leaf = (len(self.arr) // 2) - 1
    
        for i in range(last_non_leaf, -1, -1):
            self.bubble_down(i)
````

````python
# Testing
MyHip = MaxHeap([1,2,30,4,5])
print(MyHip.peek_max())
MyHip.insert(10)
MyHip.delete_max()
print(MyHip.peek_max())
````

# Sorting Comparison:
### Both Merge Sort and Heap Sort are O(nlogn))
### Quick Sort > Merge Sort > Heap Sort
### Quick sort advantage -> unlike merge sort, is an in-place algorithm, so better memory management
### Quick sort disadvantage -> worst case time is worse than heap sort and merge sort (for example for sorted list)

# Quick Sort

````python
def partition(A, left, right):
    leftp = left
    rightp = right - 1
    pivot = A[right]

    while leftp <= rightp:
        while leftp <= rightp and A[leftp] <= pivot:
            leftp += 1

        while leftp <= rightp and pivot < A[rightp]:
            rightp -= 1

        if leftp < rightp:
            A[rightp], A[leftp] = A[leftp], A[rightp]

    A[leftp], A[right] = A[right], A[leftp]
    return leftp
        
def quick_sort(A,left=0, right=len(A)-1):
    if(right<=left):
        return
    index = partition(A,left,right)
    quick_sort(A,left, index-1)
    quick_sort(A,index+1,right)
    

    
````

````python
A = [2,4,8,1,3,7,3,6,4.5]
partition(A,0,len(A)-1)
print(A)
````

````python
quick_sort(A)
print(A)
````
