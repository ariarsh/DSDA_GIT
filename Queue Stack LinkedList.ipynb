# Queue Stack LinkedList

```python
class Queue:
    def __init__(self, max_size):
        self.max_size = max_size
        self.Q = [0] * self.max_size
        self.num = 0
        self.first = 0

    def enqueue(self, item):
        if self.num >= self.max_size:
            raise Exception("Queue is full")
        self.Q[(self.first + self.num) % self.max_size] = item
        self.num += 1

    def dequeue(self):
        if (self.num == 0):
            raise Exception("Nothing to dequeue")
        item = self.Q[self.first]
        self.first = (self.first + 1) % self.max_size
        self.num -= 1
        return item
        
    def front(self):
        if (self.num == 0):
            raise Exception("Empty")
        return self.Q[self.first]

    def is_empty(self):
        return self.num
            
    def is_full(self):
        return self.num>=self.max_size
            
    def size(self):
        return self.num
    
    def print(self):
        for i in range(self.num):
            print(self.Q[(self.first + i) % self.max_size], end=" ")
        print()
```

```python
# Test
a=queue(10)
a.enqueue("Arya")
a.enqueue("Ali")
a.enqueue("Sib")
a.dequeue()
a.print()
```

```python
# Stack
class stack:
    def __init__(self,max_size):
        self.max_size = max_size
        self.stack = [0]*max_size
        self.num = 0

    def push(self,item):
        if(self.num>=self.max_size):
            raise Exception("Full")
        self.stack[self.num]=item
        self.num +=1

    def pop(self):
        if(self.num==0):
            raise Exception("Empty")
        item = self.stack[self.num-1]
        self.num -=1
        return item

    def top(self):
        if(self.num==0):
            raise Exception("empty stack")
        return self.stack[self.num-1]

    def print(self):
        for i in range(self.num):
            print(self.stack[i], end =" ")
        
```

```python
s=stack(10)
s.push("Arya")
s.push("Joe")
s.push("Sina")
s.pop()
s.print()
```

```python
# Linked List
#insert->O(1)
#remove->O(1)
#seach->O(n)
class Node:
    def __init__(self, data):
        self.data = data  
        self.next = None  
        self.prev = None  
    
class List:
    def __init__(self):        
        self.head = Node(None)
        self.head.next = self.head
        self.head.prev = self.head
        self.n = 0
        
    def get(self,ind):
        if ind >= self.size() : 
            raise Exception("Out of list")
        x = self.head.next
        for i in range(ind) :
            x = x.next
        return x
    
    def insert_after(self, x, data):
        y = Node(data)
        self.n += 1
        y.prev = x
        y.next = x.next
        x.next = y
        y.next.prev = y
        return y
        
    def delete(self, x):
        self.n -= 1
        x.prev.next = x.next
        x.next.prev = x.prev
        return x
        
    def find(self, val):
        x = self.head.next
        for i in range(self.size()) :
            if x.data == val :
                return x
            x = x.next
        return None
    
    def size(self):
        return self.n
      
    def is_empty(self):
        return self.n==0
```
