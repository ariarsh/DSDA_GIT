# Sorting

```python
# Bubble Sort
A=[7,5,1, 6, 9, 10,4,2,1]
def bubble_sort(A):
    for i in range(len(A)):
        for j in range(len(A)-i-1):
            if(A[j]>=A[j+1]):
                A[j],A[j+1] = A[j+1], A[j]
    return A

print(bubble_sort(A))
```

```python
# Insertion Sort
A=[7,5,1, 6, 9, 10,4,2,1]
def insertion_sort(A):
    for i in range(1,len(A)):
        insertion=A[i]
        for j in range(i-1,-1,-1):
            if(A[j]>insertion):
                A[j+1]=A[j]
                A[j]=insertion
            else:
                A[j+1]=insertion
                break
    return A
print(insertion_sort(A))
                
```

```python
# Selection Sort
A=[7,5,1, 6, 9, 10,4,2,1]
def selection_sort(A):
    for i in range(len(A)):
        selection=i
        for j in range(i,len(A)):
            if(A[j]<=A[selection]):
               selection=j
        A[i],A[selection]=A[selection],A[i]
        
    return A
print(selection_sort(A))
            
```

```python
#Buccket Sort
A=[7,5,1, 6, 9, 10,4,2,30,1]
def bucket_sort(A):
    res = A[:]
    max_num = max(res)
    bucket_num = max_num // 10 + 1
    buckets = [[] for i in range(bucket_num)]
    
    for i in res:
        buckets[i // 10] += [i]

    for i in range(bucket_num):
        buckets[i] = selection_sort(buckets[i])

    res = []
    for i in range(bucket_num):
        res += buckets[i]
    
    return res
print(bucket_sort(A))
```

```python
#Merge Sort
A=[7,5,1, 6, 9, 10,4,2,30,1]

def merge_sort(A):
    if(len(A)<2):
        return A
    mid=len(A)//2
    Left = A[:mid]
    Right = A[mid:]
    Left = merge_sort(Left)
    Right = merge_sort(Right)
    A=[]
    i=0
    j=0
    while i<len(Left) and j<len(Right):
        if(Left[i]<=Right[j]):
            A.append(Left[i])
            i+=1
        elif(Left[i]>Right[j]):
            A.append(Right[j])
            j+=1
    A+=Left[i:]
    A+=Right[j:]
    return A
print(merge_sort(A))     
```
